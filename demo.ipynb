```python
import subprocess
import time
import os

# Ensure uv is installed
!curl -LsSf https://astral.sh/uv/install.sh | sh

# Change to the project directory to ensure 'uv sync' and 'uvicorn' can find their files
repo_dir = "vllm-ops-engine"
if not os.path.exists(repo_dir) and not os.getcwd().endswith(repo_dir):
    print(f"Error: Directory '{repo_dir}' not found. Please ensure the repository is cloned correctly.")
else:
    if os.path.exists(repo_dir):
        os.chdir(repo_dir)
    print(f"Current working directory: {os.getcwd()}")

    # Install dependencies using uv sync
    print("Installing dependencies with uv sync...")
    !uv sync
    print("Dependencies installed.")

    # Start the FastAPI backend server in the background using the correct uv path
    print("Starting FastAPI backend server...")
    server = subprocess.Popen([
        "uv", "run", "uvicorn", "src.api:app", "--host", "0.0.0.0", "--port", "8000"
    ])

    # Give it a few seconds to load the sentence-transformers and vLLM models
    time.sleep(20)
    print("Server started in the background!")
```
